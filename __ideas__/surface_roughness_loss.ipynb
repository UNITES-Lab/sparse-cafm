# **Surface Roughness Loss**
***

Currently, we train our super-resolution models for topological upsampling via a standard L1 loss.

## Standard L1

$$
\mathcal{L}(\theta) = ||y - \hat{y}||
$$

However, we would like our model to produce outputs with ***similar surface roughness*** to the ground truth data. The equation for RMSE surface roughness is simply a scaled and shifted RSME error where...

$\alpha = mean(X)$

$\beta = 1e9$

$$Sr(X) = \beta * \sqrt{(X - \alpha)^2}$$

This formula is differentiable, therefore we can easily extend the standard L1 to include a penalty term for surface roughness...

## L1 + Surface Roughness Penalty

$$
\mathcal{L}(\theta) = ||y - \hat{y}|| + \lambda || Sr(y) - Sr(\hat{y}) ||
$$

where $\lambda$ is a hyperparameter.

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

# Create a figure and GridSpec layout
fig = plt.figure(figsize=(50,20))
gs = GridSpec(2, 3, figure=fig, width_ratios=[1, 1, 1], height_ratios=[1, 1], wspace=0.05, hspace=0.05)

# Create subplots
axs = [
    fig.add_subplot(gs[0:2, 0]),  # Left column (spanning two rows)
    fig.add_subplot(gs[0, 1]),    # Top-right
    fig.add_subplot(gs[0, 2]),    # Top-right
    fig.add_subplot(gs[1, 1]),    # Bottom-right
    fig.add_subplot(gs[1, 2]),    # Bottom-right
]

images_and_titles = [
    (img_gp,   "Gaussian Process Regression"),
    (img_bic,  "Bicubic"),
    (img_lin,  "Linear"),
    (img_nn,   "Nearest Neighbor"),
    (img_ours, "Our Method"),
]

# Iterate over images and axes
for i, (ax, (image, title)) in enumerate(zip(axs, images_and_titles)):
    ax.imshow(image)
    ax.set_title(f"{chr(97 + i).upper()}. {title}", fontsize=18, weight='bold', loc='left')
    ax.axis('off')

# Display the figure and save it
plt.show()
```

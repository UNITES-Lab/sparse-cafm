```python
import pandas as pd
fp = r"/playpen/mufan/levi/tianlong-chen-lab/material-super-resolution/__exps__/y-task-formulations/6. p(y | y_sparse)/6b. evals/2025-02-03_14-26-00__swinir->unet.out/results.csv"
df = pd.read_csv(fp)
df.head()
```

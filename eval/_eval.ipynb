```python
import pandas as pd
fp = r"/playpen/mufan/levi/tianlong-chen-lab/material-super-resolution/__exps__/y-task-formulations/p(y | y_sparse)/6c. surrogate-train-runs/2025-02-02_06-10-31_surrogate-depth=8/results.csv"
df = pd.read_csv(fp)
df.head()
```

```python
summary_stats = {
    "MAE": df["mae"].mean(),
    "MSE": df["mse"].mean(),
    "PSNR": df["psnr"].mean() - 6.02, # HACK: scale PSNR [-1, 1] -> [0, 1]
    "SSIM": df["ssim"].mean(),
    "OLDER": df["older"].mean(),
}
summary_str = "\n".join([f"{key}: {value:.6f}" for key, value in summary_stats.items()])
print("Stats:\n" + summary_str)
```

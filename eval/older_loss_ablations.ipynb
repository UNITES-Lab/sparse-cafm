```python
import pandas as pd
from glob import glob

evals_dir = "/playpen/mufan/levi/tianlong-chen-lab/material-super-resolution/__exps__/y-task-formulations/p(y | y_sparse)/b. evals"
dirs = glob(f"{evals_dir}/*02-19*")
dirs.sort()
```

```python
import matplotlib.pyplot as plt

exp_names = [s.split("02-19_")[-1] for s in dirs]; exp_names
exp_results = [glob(d + "/*.csv")[0] for d in dirs]; exp_results

for name, result in zip(exp_names, exp_results):
    assert name in result
    df = pd.read_csv(result)
    summary_stats = {
    "MAE": df["mae"].mean(),
    "MSE": df["mse"].mean(),
    "PSNR": df["psnr"].mean(),
    "SSIM": df["ssim"].mean(),
    "OLDER": df["older"].mean(),
    }
    summary_str = "\n".join([f"{key}: {value:.6f}" for key, value in summary_stats.items()])
    print(f"{name} stats:\n" + summary_str)
    print("------------------------------")
```

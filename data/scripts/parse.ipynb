```python
import numpy as np

fp_y = "/Users/leviharris/Desktop/research/tianlong-chen-lab/8-27-24-initial-assignment/microscopy-data/mm_20170124.095_512x512_MoS2_txt"
fp_x = "/Users/leviharris/Desktop/research/tianlong-chen-lab/8-27-24-initial-assignment/microscopy-data/mm_20170124_2_512x512_MoS2_txt"

def convert_txt_np_array(fp: str):
    
    raw_file = open(fp, 'r').readlines()

    # 2d-arr of strings
    img_parsed_str = [line.split('\t') for line in raw_file]
    
    # clean up array and convert to np format
    img_parsed_float = []
    
    for subarr in img_parsed_str:
        subarr_float = []
        for val in subarr:
            # remove newline chars
            subarr_float.append(float(val.replace("\n", "")))
        img_parsed_float.append(subarr_float)
        
    # (H, W, D)
    img_parsed_float = np.array(img_parsed_float)
    return img_parsed_float

arr_x = convert_txt_np_array(fp_x)
arr_y = convert_txt_np_array(fp_y)
arr_x.shape, arr_y.shape
```

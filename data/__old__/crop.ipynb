```python
import os
import numpy as np
from PIL import Image
from glob import glob

image_dir = "/playpen/mufan/levi/tianlong-chen-lab/material-super-resolution/data/full-sized-toy-dataset"
all_img_fps = glob(f"{image_dir}/*/*.png")
ex_fp = all_img_fps[0]
```

```python
OG_HW = 512
HEIGHT_OFFSET = 35
WIDTH_L_OFFSET = 30
WIDTH_R_OFFSET = 90

dst_dir = "/playpen/mufan/levi/tianlong-chen-lab/material-super-resolution/data/full-sized-toy-dataset-cropped"

img = Image.open(ex_fp)
cropped_img = img.crop((0 + WIDTH_L_OFFSET, 0 + HEIGHT_OFFSET, OG_HW - WIDTH_R_OFFSET, OG_HW - HEIGHT_OFFSET))
resized_img = cropped_img.resize((OG_HW, OG_HW))

# for each image in all_img_fps
# 1. crop + resize img
# 2. save to dst_dir
# 3. create new folders as needed

for fp in all_img_fps:
    img = Image.open(fp)
    cropped_img = img.crop((0 + WIDTH_L_OFFSET, 0 + HEIGHT_OFFSET, OG_HW - WIDTH_R_OFFSET, OG_HW - HEIGHT_OFFSET))
    resized_img = cropped_img.resize((OG_HW, OG_HW))
    dst_fp = fp.replace(image_dir, dst_dir)
    os.makedirs(os.path.dirname(dst_fp), exist_ok=True)
    resized_img.save(dst_fp)
```

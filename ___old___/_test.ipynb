```python
import torch
import tensorhue as th

og_img = torch.rand((1, 3, 256, 256))
pred_img = torch.rand((1, 3, 256, 256))
mask = (torch.rand((1, 1, 256, 256)) > 0.5).to(torch.bool)
mask = mask.repeat(1, 3, 1, 1); mask.shape
```

```python
given_pixels = og_img * mask
pred_pixels = pred_img * ~mask
final_prediction = given_pixels + pred_pixels
final_prediction.shape
torch.nn.functional.l1_loss(final_prediction, og_img)
```

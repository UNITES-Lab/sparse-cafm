```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

# --- Figure 1: Surface Height Colorbar ---
fig1 = plt.figure(figsize=(1.5, 7), dpi=600)
ax1 = fig1.add_axes([0.4, 0.0, 0.40, 1.0])  # This positioning is correct

# Remove the default padding around the figure
fig1.subplots_adjust(left=0, right=1, top=1, bottom=0)

cmap = mpl.cm.viridis
norm1 = mpl.colors.Normalize(vmin=0.25, vmax=2.00)
cb1 = mpl.colorbar.ColorbarBase(ax1, cmap=cmap, norm=norm1, orientation='vertical')
cb1.set_label('Surface Height (nm)', rotation=270, labelpad=50, fontsize=24)
ticks1 = np.arange(0.25, 2.01, 0.25)
cb1.set_ticks(ticks1)
cb1.set_ticklabels([f'{t:.2f}' for t in ticks1])
cb1.ax.tick_params(labelsize=22)
plt.show()

# --- Figure 2: Surface Current Colorbar ---
fig2 = plt.figure(figsize=(1.5, 7), dpi=600)
ax2 = fig2.add_axes([0.4, 0.0, 0.40, 1.0]) # This positioning is correct

# Remove the default padding around the figure
fig2.subplots_adjust(left=0, right=1, top=1, bottom=0)

norm2 = mpl.colors.Normalize(vmin=0.2, vmax=1.2)
cb2 = mpl.colorbar.ColorbarBase(ax2, cmap=cmap, norm=norm2, orientation='vertical')
cb2.set_label('Surface Current (nA)', rotation=270, labelpad=50, fontsize=24)
ticks2 = np.round(np.arange(0.2, 1.21, 0.2), 1)
cb2.set_ticks(ticks2)
cb2.set_ticklabels([f'{t:.1f}' for t in ticks2])
cb2.ax.tick_params(labelsize=22)
plt.show()
```

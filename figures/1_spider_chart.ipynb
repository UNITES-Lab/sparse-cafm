```python
import matplotlib.pyplot as plt
import numpy as np

# Define the feature labels
labels = np.array(['Average Surface Current', 'Length of Grain Boundaries', 'Coverage %'])
num_vars = len(labels)

# Define the data for each group
theirs = [50, 30, 20]
ours = [70, 80, 90]

# Append the first value at the end to close the loop
ours += ours[:1]
theirs += theirs[:1]

# Calculate the angles for each axis (in radians)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

# Create a polar subplot with increased DPI for high resolution
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True), dpi=300)

# Plot "ours" data with a label for the legend
ax.plot(angles, ours, color='red', linewidth=2, label='ours')
ax.fill(angles, ours, color='red', alpha=0.25)

# Plot "theirs" data with a label for the legend
ax.plot(angles, theirs, color='blue', linewidth=2, label='theirs')
ax.fill(angles, theirs, color='blue', alpha=0.25)

# Add labels to each axis at the proper angles
ax.set_thetagrids(np.degrees(angles[:-1]), labels, fontsize=12)

# Set a title for the chart with increased font size
ax.set_title('Electric Characterization % Accuracy', y=1.1, fontsize=14)

# Enable grid lines for clarity and improve layout
ax.grid(True)
plt.tight_layout()

# Add a legend outside the plot area for clarity
ax.legend(loc='upper right', bbox_to_anchor=(1.1, 1.1), fontsize=12)

# Display the plot
plt.show()
```

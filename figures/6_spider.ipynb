```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker


df = pd.DataFrame(rows)
metrics = [
    "Mean Surface Current",
    "Coverage Percent",
    "Length Detected Curves",
    "Area Circular Shapes",
    "Area Extended Shapes",
    "Total Defect Area",
]

for m in metrics:
    if 0 in df[m].values:
        min_positive = df.loc[df[m] > 0, m].min()
        df.loc[df[m] == 0, m] = min_positive * 0.1

# wide -> long
long_df = df.melt(
    id_vars=["Dataset", "Ratio", "Input"],
    value_vars=metrics,
    var_name="Metric",
    value_name="Score",
)
pivoted_df = long_df.pivot_table(
    index=["Dataset", "Ratio", "Metric"], columns="Input", values="Score"
).reset_index()

# calculate %improvement; higher=better
pivoted_df["Improvement"] = (
    (pivoted_df["Sparse"] - pivoted_df["Predicted"]) / pivoted_df["Sparse"] * 100
)

datasets = pivoted_df["Dataset"].unique()
ratios = sorted(pivoted_df["Ratio"].unique())
metric_order = metrics

fig, axes = plt.subplots(
    nrows=1, ncols=len(datasets),
    sharey=True, dpi=600,
    figsize=(5.5, 7),
    squeeze=False          # <- fix
)
axes = axes.flatten()
fig.suptitle("Scorecard: SparseC-AFM predictions vs. sparse maps", fontsize=16, y=0.93)

vmax = pivoted_df["Improvement"].max()
vmin = -vmax

for ax, dataset in zip(axes, datasets):

    subset = pivoted_df[pivoted_df["Dataset"] == dataset]
    heatmap_data = subset.pivot(index="Metric", columns="Ratio", values="Improvement")
    heatmap_data = heatmap_data.reindex(metric_order)

    # ── draw the image ───────────────────────────────────────────────
    im = ax.imshow(
        heatmap_data,
        cmap=cmap,
        vmin=vmin, vmax=vmax,
        aspect="auto",
        interpolation="nearest"   # keeps edges crisp
    )

    n_rows, n_cols = heatmap_data.shape
    ax.set_xticks(np.arange(n_cols + 1) - 0.5, minor=True)
    ax.set_yticks(np.arange(n_rows + 1) - 0.5, minor=True)
    ax.grid(which="minor", color="black", linewidth=0.2)
    ax.tick_params(which="minor", length=0)  # hide minor tick marks

    ax.set_title(dataset, fontsize=12)
    ax.set_xlabel("Data Reduction (×)", fontsize=12)
    ax.set_xticks(np.arange(len(ratios)))
    ax.set_xticklabels(ratios)

# axes[0].set_ylabel("Error Metric")
axes[0].set_yticks(np.arange(len(metric_order)))

for i, metric in enumerate(metric_order):
    metric_order[i] = metric_order[i].lower()

metric_order[0] = "mean surface\n current"
metric_order[1] = "converage\n percent"
metric_order[2] = "length detected\n curves"
metric_order[3] = "area circular\n shapes"
metric_order[4] = "area extended\n shapes"
metric_order[5] = "total defect\n area"

axes[0].set_yticklabels(metric_order, fontsize=10)

for label in axes[0].get_yticklabels():
    label.set_rotation(30)

fig.tight_layout(rect=[0, 0, 0.85, 0.95])

# add cbar
cbar_ax = fig.add_axes([0.86, 0.085, 0.055, 0.792])
cbar = fig.colorbar(im, cax=cbar_ax)
cbar.set_label("Percent better than baseline", rotation=90, labelpad=10, fontsize=12)
cbar.ax.yaxis.set_major_formatter(mticker.PercentFormatter())

plt.show()
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker


df = pd.DataFrame(rows)

metrics = [
    "Mean Surface Current",
    "Coverage Percent",
    "Length Detected Curves",
    # "Area Circular Shapes",      # dropped  (index 3)
    # "Area Extended Shapes",      # dropped  (index 4)
    "Total Defect Area",
]


# replace zeros with a tiny positive value (10 % of the min positive in that column)
for m in metrics:
    if 0 in df[m].values:
        min_positive = df.loc[df[m] > 0, m].min()
        df.loc[df[m] == 0, m] = min_positive * 0.1

# long form
long_df = df.melt(
    id_vars=["Dataset", "Ratio", "Input"],
    value_vars=metrics,
    var_name="Metric",
    value_name="Score",
)

# pivot so we can compute improvement
pivoted_df = (
    long_df
      .pivot_table(index=["Dataset", "Ratio", "Metric"], columns="Input", values="Score")
      .reset_index()
)

pivoted_df["Improvement"] = (
    (pivoted_df["Sparse"] - pivoted_df["Predicted"]) / pivoted_df["Sparse"] * 100
)

# ── 2) Prepare figure  ────────────────────────────────────────────────
datasets     = pivoted_df["Dataset"].unique()
ratios       = sorted(pivoted_df["Ratio"].unique())
metric_order = metrics

vmax = pivoted_df["Improvement"].max()
vmin = -vmax

fig, axes = plt.subplots(
    nrows=1, ncols=len(datasets),
    sharex=True,
    dpi=600, figsize=(5.5, 4.5),
    squeeze=False
)
axes = axes.flatten()
fig.suptitle("Scorecard: SparseC-AFM predictions vs. sparse maps", fontsize=13, y=0.91)


# pretty metric labels for the x-axis (keep the same order)
short_metric_labels = [
    "mean surface\ncurrent",
    "coverage\npercent",
    "ground boundary\nlength",
    # "area circular\nshapes",     # dropped
    # "area extended\nshapes",     # dropped
    "total defect\narea",
]

for ax, dataset in zip(axes, datasets):

    subset = pivoted_df[pivoted_df["Dataset"] == dataset]

    heatmap_data = (
        subset
          .pivot(index="Ratio", columns="Metric", values="Improvement")
          .reindex(columns=metric_order)        # keep metric order consistent
          .sort_index(ascending=True)           # ratios top→bottom (4×, 8×, 16× …)
    )

    im = ax.imshow(
        heatmap_data,
        cmap=cmap, vmin=vmin, vmax=vmax,
        aspect="auto", interpolation="nearest"
    )

    # minor grid lines
    n_rows, n_cols = heatmap_data.shape
    ax.set_xticks(np.arange(n_cols + 1) - 0.5, minor=True)
    ax.set_yticks(np.arange(n_rows + 1) - 0.5, minor=True)
    ax.grid(which="minor", color="gray", linewidth=0.2)
    ax.tick_params(which="minor", length=0)

    # major ticks
    ax.set_xticks(np.arange(n_cols))
    ax.set_xticklabels(short_metric_labels, rotation=35, ha="right", fontsize=8)

    ax.set_yticks(np.arange(n_rows))
    ax.set_yticklabels(heatmap_data.index, fontsize=10)

    # titles & axis labels
    ax.set_title(dataset, fontsize=12)
    if ax is axes[0]:
        ax.set_ylabel("Data Reduction (×)", fontsize=10)

fig.tight_layout(rect=[0, 0, 0.86, 0.95])

cbar_ax = fig.add_axes([0.87, 0.210, 0.045, 0.634])
cbar    = fig.colorbar(im, cax=cbar_ax)
cbar.set_label("Percent better than baseline", rotation=90, labelpad=6, fontsize=10)
cbar.ax.tick_params(axis='y', labelsize=8)   # ← make them smaller (e.g. 8 pt)
cbar.ax.yaxis.set_major_formatter(mticker.PercentFormatter())

plt.show()
```

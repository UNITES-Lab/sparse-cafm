```python
import numpy as np
import matplotlib.pyplot as plt

training_samples = np.array([0, 1, 2, 3])
psnrs = np.array([51.32, 51.57, 51.99, 52.13])

fig, ax = plt.subplots(figsize=(4, 3), dpi=600)
ax.plot(training_samples, psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, color='#4B9CD3')
ax.set_xlabel('# Full-Resolution Training Samples', fontsize=10)
ax.set_ylabel('PSNR (dB)', fontsize=10)
ax.grid(False)
ax.tick_params(axis='both', labelsize=10)
ax.set_xticks(training_samples)

plt.grid(False)
plt.tight_layout()
plt.show()
```

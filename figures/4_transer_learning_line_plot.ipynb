```python
import matplotlib.pyplot as plt
import numpy as np

training_samples = np.array([0, 1, 2, 3])

# 2x
iix_psnrs = np.array([51.32, 51.57, 51.99, 52.13])
# 4x
ivx_psnrs = np.array([40.89, 44.89, 45.89, 49.89])
# 8x
xiii_psnrs = np.array([32.82, 32.89, 36.89, 40.89])

fig, ax = plt.subplots(figsize=(4, 3), dpi=600)

ax.plot(training_samples, iix_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, label='2x')
ax.plot(training_samples, ivx_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, label='4x')
ax.plot(training_samples, xiii_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, label='8x')

ax.set_xlabel('# Full-Resolution Training Samples', fontsize=10)
ax.set_ylabel('PSNR (dB)', fontsize=10)
ax.set_xticks(training_samples)
ax.tick_params(axis='both', labelsize=10)

legend = ax.legend(
    title="Upsampling", 
    loc="lower right", 
    title_fontsize=8
)

ax.grid(True)
plt.tight_layout()
plt.show()
```

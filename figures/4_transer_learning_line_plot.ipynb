```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np


training_samples = np.array(["No Pre-Train", 0, 1, 2, 3])

# 2x
iix_psnrs = np.array([26.18, 51.32, 51.57, 51.99, 52.13])
# 4x
ivx_psnrs = np.array([19.78, 40.89, 40.87, 40.95, 40.88])
# 8x
xiii_psnrs = np.array([17.05, 32.82, 32.23, 33.12, 32.78])

fig, ax = plt.subplots(figsize=(4, 3), dpi=600)

# Use hex colors from palette
ax.plot(training_samples, iix_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, color='#4B9CD3', label='2x')
ax.plot(training_samples, ivx_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, color='#13294B', label='4x')
ax.plot(training_samples, xiii_psnrs, marker='o', linestyle='-', linewidth=2, markersize=8, color='#151515', label='8x')

ax.set_xlabel('# Full-Resolution Training Samples', fontsize=10)
ax.set_ylabel('PSNR (dB)', fontsize=10)
ax.set_xticks(training_samples)
ax.tick_params(axis='both', labelsize=8)

legend = ax.legend(title="Upsampling", loc="lower right", title_fontsize=8)

ax.grid(True)
plt.tight_layout()
plt.show()
```
